### JSON & API Basics

```python
import json

# JSON-like structure
agent_data = {
    "agent": "Alpha",
    "role": "assistant",
    "goals": ["plan", "analyse", "act"]
}
```

```python
# Convert to JSON string
json_str = json.dumps(agent_data)
print("JSON string:", json_str)
```

```python
# Back to Python object
parsed = json.loads(json_str)
print("Agent goaols:", parsed["goals"])
```

### Mini- Project: Agent To do list

```python
tasks = []

def add_task(task):
    tasks.append(task)


def list_tasks():
    for i, t in enumerate(tasks, 1):
        print(f"{i}. {t}")


add_task("Collect user query")
add_task("Call LLM")
add_task("Return result")

list_tasks()
```

```python
# Add a remove task function
def remove_task(index):
    if 0 <= index < len(tasks):
        removed = tasks.pop(index)
        print(f"Removed task: {removed}")
    else:
        print("Invalid task index.")

remove_task(1)  # Remove the second task
list_tasks()  # List tasks after removal
```

```python
# Save the list to a file as JSON
with open("tasks.json", "w") as f:
    json.dump(tasks, f)

# Read the tasks back from the file
with open("tasks.json", "r") as f:
    loaded_tasks = json.load(f)
    print("Loaded tasks from file:")
    for t in loaded_tasks:
        print(t)
```
